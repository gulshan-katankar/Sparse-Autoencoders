```python
from transformer_lens import HookedTransformer

model = HookedTransformer.from_pretrained(
    "gpt2-small",
    device="cuda"
)
```

**RUN THE EXPERIMENT**

```python
text = "The capital of France is"

logits = model(text)

print(logits.shape)
```

```python
next_token = logits[0, -1].argmax()

print(model.tokenizer.decode([next_token]))
```

**EXTRACTNG THE INTERNAL ACTIVATIONS**

```python
logits, cache = model.run_with_cache(text)
```

```python
print(cache.keys())
```
